```python
'''
Q4 Regional Sales Performance

Calculate the total sales generated by each region.

Then identify:

Which region generated the highest revenue?'''

```

```python
#Calculate the total sales generated by each region.
df.groupby('Region')['Sales'].sum()
```

```python
#hich region generated the highest revenue?
df.groupby('Region')['Sales'].sum().sort_values(ascending=False).head(1)
```

so the region **East** is generating the highest sales among all the regions


```python
''' Q5 Product Performance Analysis

Calculate the total quantity sold for each product.

Which product appears to be the most popular?'''
```

```python
# Calculate the total quantity sold for each product.
df.groupby('Product')['Quantity'].sum()
```

```python
# Which product appears to be the most popular
df.groupby('Product')['Quantity'].sum().sort_values(ascending=False).head(1)
```

so the product Monitor is most popular with the highest quantity sold i.e 156

```python
'''Q6 Multi-Level Grouping

Group the dataset by: Region and Category'''

df.groupby(['Region','Category'])['Sales'].sum()

#Calculate the total sales for each combination.
df.groupby(['Region','Category'])['Sales'].sum()

#What insights can you derive from this analysis?
df.groupby(['Region','Category'])['Sales'].sum()
```

from the above result it can be easily interpreted thatregion east is performing good in comparison to other regions
also in terms of categories, electronics is the leading performer in all the regions followed by Accessories and Office **bold text**

```python
'''
Q7 Pivot Table Creation

Create a pivot table showing:

Rows → Region

Columns → Product

Values → Total Sales

Which region sells the most laptops?
answer is east region

'''

pivot_df = df.pivot_table(
    index='Region',
    columns='Product',
    values='Sales',
    aggfunc='sum'
)



print(pivot_df)
```
