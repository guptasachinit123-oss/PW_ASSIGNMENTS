```python
# univariate analysis of number of bedroom column

bedroom_counts = df['number of bedrooms'].value_counts()
print(bedroom_counts)
plt.figure(figsize=(10, 6))
plt.bar(bedroom_counts.index, bedroom_counts.values, color='skyblue')
plt.xlabel('Bedrooms')
plt.ylabel('Count')
plt.title('Distribution of Bedrooms')
plt.show()
```

```python
columns_to_analyze = ['Price', 'number of bedrooms', 'number of bathrooms', 'number of floors', 'waterfront present']
corr_matrix = df[columns_to_analyze].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('All-in-One Bivariate Analysis (Correlation Heatmap)')
plt.show()
```

```python
# Q1. What is the average, median, and standard deviation of house prices?
answer
average_price = df['Price'].mean()
median_price = df['Price'].median()
std_dev_price = df['Price'].std()
print(f"Average Price: {average_price}")
print(f"Median Price: {median_price}")
print(f"Standard Deviation of Price: {std_dev_price}")
```

```python
# Q2.Which number of bedrooms is most common?
bedroom_counts = df['number of bedrooms'].value_counts()
most_common_bedroom = bedroom_counts.idxmax()
print(f"The most common number of bedrooms is {most_common_bedroom}.")
```

```python
#. Q3.Check if the price data is skewed.
#answer
skewness = df['Price'].skew()
if skewness > 0:
    print("The price data is positively skewed.")
elif skewness < 0:
    print("The price data is negatively skewed.")
else:
    print("The price data is approximately symmetric.")
```

```python
#Q4. What is the average price for each number of bedrooms?
#answer
avg_price_per_bedroom = df.groupby('number of bedrooms')['Price'].mean().reset_index()
print(avg_price_per_bedroom)
```

```python
QQ6. Identify any anomalies where houses have high prices but low areas.
#answer
high_price_limit = df['Price'].quantile(0.95)
low_area_limit = df['Area of the house(excluding basement)'].quantile(0.25)
nomalies_df = df[(df['Price'] >= high_price_limit) & (df['Area of the house(excluding basement)'] <= low_area_limit)]
low_area_limit = df['Area of the house(excluding basement)'].quantile(0.25)
anomalies_df = df[(df['Price'] >= high_price_limit) & (df['Area of the house(excluding basement)'] <= low_area_limit)]

print(anomalies_df[['Price', 'Area of the house(excluding basement)']].head())

```

```python
#Q7. Compare average price based on number of floors and houses with or without waterfront.
#answer
omparison_table =df.groupby(['number of floors', 'waterfront present'])['Price'].mean().reset_index()
print(omparison_table)
```

```python
#Q8. Identify the minimum and maximum house price. What does this indicate?
#answer
min_price = df['Price'].min()
max_price = df['Price'].max()
print(f"The minimum house price is: {min_price}")
print(f"The maximum house price is: {max_price}")
price_range = max_price - min_price

print(f"Minimum House Price (price of the cheapest house): {min_price:,}")
print(f"Maximum House Price (price of the costliest house): {max_price:,}")
print(f"Total Price Range ( price difference) : {price_range:,}")
```

```python
#Q9. Which location (zipcode/area) has the highest average price?
#answer
highest_avg_zip = df.groupby('Postal Code')['Price'].mean().idxmax()
highest_avg_value = df.groupby('Postal Code')['Price'].mean().max()
print(f"The location with the highest average price is {highest_avg_zip} with an average price of {highest_avg_value:}.")
```
