 **output**:- DECLARE @i INT = 1;
WHILE @i <= 10000
BEGIN
    INSERT INTO SalesData (Name, Age, City, PurchaseAmount, PurchaseDate)
    VALUES (
        'Customer_' + CAST(@i AS VARCHAR(10)),
        FLOOR(RAND()*(65-18+1))+18, -- Random age between 18 and 65
        CHOOSE(FLOOR(RAND()*5)+1, 'New York', 'Los Angeles', 'Chicago', 'Houston', 'Miami'),

    );
    SET @i = @i + 1;
END;

FROM SalesData

**task B-**

**output**:-

FROM SalesData
GROUP BY City

**Question 10** : Generate VBA code using Copilot for cleaning data (removing duplicates, filling null values,
format table).

**answer10:-**
prompt:-
Write a VBA macro that cleans a table by removing duplicate rows, replacing empty cells with 'N/A', and auto-fitting the table columns.

**output**:-

Sub CleanDataMacro()
    Dim ws As Worksheet
    Set ws = ActiveSheet
    Dim rng As Range
    Set rng = ws.UsedRange

    ' 1. Remove Duplicates
    rng.RemoveDuplicates Columns:=Evaluate("ROW(1:" & rng.Columns.Count & ")"), Header:=xlYes

    ' 2. Fill missing null values with N/A
    On Error Resume Next
    rng.SpecialCells(xlCellTypeBlanks).Value = "N/A"
    On Error GoTo 0

    ' 3. Format Table (Auto-fit columns)
    rng.Columns.AutoFit

    MsgBox "Data cleaning complete!", vbInformation
End Sub
