```python
#answer3:-

quantity_mean = df['Quantity'].mean()
print(quantity_mean)
df['Quantity'].fillna(df['Quantity'].mean(),inplace=True)
df.isnull().sum()

df['Price'].fillna(df['Price'].mean(),inplace=True)
df.isnull().sum()
```

```python
#answer4:-

df.dropna(subset=['Product Category', 'Region'])

df.isnull().sum()
```

```python
#answer5:-

df['Revenue']=np.multiply(df['Quantity'].values,df['Price'].values)
df.head(5)
```

```python
#answer6:-

total_revenue = np.sum(df['Revenue'].values)
print("total revenue generated i thr datast is",total_revenue)
```

```python
#answer7:-
category_revenue = df.groupby('Product Category')['Revenue'].sum()
print("--- Total Revenue by Product Category --->>",category_revenue)
```

```python
#answer8:-

# to find the top3 product category firter sort in ascending orderso that the top product categories comes at the top
sorted_categories = category_revenue.sort_values(ascending=False)

# now to find top3 categories
print("Top 3 Product Categories are ---->>>>")
sorted_categories.head(3)

print("now the last  3 Product Categories are ---->>>>")
sorted_categories.tail(3)
```

```python
#nswer9: -
region_revenue = df.groupby('Region')['Revenue'].sum()

#to find the region with the highest revenue
region_revenue.idxmax()
print(region_revenue.idxmax())

#to find the region with the lowest revenue
region_revenue.idxmin()
print(region_revenue.idxmin())
```
