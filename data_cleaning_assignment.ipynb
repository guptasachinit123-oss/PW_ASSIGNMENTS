```python
#answer6:-

# Calculate Q1, Q3, and IQR
Q1 = df['Treatment_Cost'].quantile(0.25)
Q3 = df['Treatment_Cost'].quantile(0.75)
#calculate iqr
IQR = Q3 - Q1

# defining boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# 3. Detect outliers
outliers = df[(df['Treatment_Cost'] < lower_bound) | (df['Treatment_Cost'] > upper_bound)]
print(f"Number of outliers detected using IQR: {len(outliers)}")
```

```python
# anser:-8
# adding a new column  that is having the log of tratment_cost values
df['Log_Treatment_Cost'] = np.log1p(df['Treatment_Cost'])

# 2. to  Compare skewness before and after
skew_before = df['Treatment_Cost'].skew()
skew_after = df['Log_Treatment_Cost'].skew()
#printing the observation
print(f"Skewness before transformation: {skew_before}")
print(f"Skewness after log transformation: {skew_after}")
```
