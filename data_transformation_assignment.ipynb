Q1. Handle Missing Values (Basic Cleaning)
Fill missing values using the following:
Age → Median
Salary → Mean
Hours_Worked_Per_Week → Median
Performance_Score → Mean
Display the dataset after handling missing values

```python
df1['Age'].fillna(df1['Age'].median(),inplace=True)



```

```python
'''Salary → Mean
Hours_Worked_Per_Week → Median
Performance_Score → Mean
Display the dataset after handling missing values'''
df1['Salary'].fillna(df1['Salary'].mean(),inplace=True)
df1['Hours_Worked_Per_Week'].fillna(df1['Hours_Worked_Per_Week'].median(),inplace=True)
df1['Performance_Score'].fillna(df1['Performance_Score'].mean(),inplace=True)
```

**Q6.** Compare Scaling Methods

MinMaxScaler

StandardScaler

On the Salary column.

```python
# creating new column of salary data to perform both scaling method

salary_data = df1[['Salary']].dropna()
```

```python
type(salary_data)
salary_data.head()
```

```python
# Applying both methods of scaling on the newly created salary column of df1 dataframe
minmax_salary = MinMaxScaler().fit_transform(salary_data)
std_salary = StandardScaler().fit_transform(salary_data)
```

```python
type(minmax_salary)
```

```python
print(minmax_salary.flatten())
```

```python
comparison_df = pd.DataFrame({
    'MinMax_Scaled_Salary': minmax_salary.flatten(),
    'Standard_Scaled_Salary': std_salary.flatten()
})
```
